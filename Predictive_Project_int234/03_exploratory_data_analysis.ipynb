# 03 - Exploratory Data Analysis


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('online_shoppers_intention.csv')
sns.set_theme(style='whitegrid')

# Revenue distribution
plt.figure(figsize=(7, 5))
sns.countplot(x='Revenue', data=df)
plt.title('Revenue Distribution')
plt.xlabel('Revenue')
plt.ylabel('Count')
plt.tight_layout()
plt.show()

# Purchase rate by Month
month_order = ['Feb', 'Mar', 'May', 'June', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
month_rate = df.groupby('Month')['Revenue'].mean().reindex(month_order).dropna()

plt.figure(figsize=(9, 5))
month_rate.plot(kind='bar')
plt.title('Purchase Rate by Month')
plt.xlabel('Month')
plt.ylabel('Purchase Rate')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Purchase rate by Visitor Type
visitor_rate = df.groupby('VisitorType')['Revenue'].mean().sort_values(ascending=False)

plt.figure(figsize=(8, 5))
visitor_rate.plot(kind='bar')
plt.title('Purchase Rate by Visitor Type')
plt.xlabel('Visitor Type')
plt.ylabel('Purchase Rate')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# Purchase rate by Weekend
weekend_rate = df.groupby('Weekend')['Revenue'].mean()

plt.figure(figsize=(7, 5))
weekend_rate.plot(kind='bar')
plt.title('Purchase Rate by Weekend')
plt.xlabel('Weekend')
plt.ylabel('Purchase Rate')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# Boxplots
fig, axes = plt.subplots(1, 3, figsize=(16, 5))
sns.boxplot(x='Revenue', y='PageValues', data=df, ax=axes[0])
axes[0].set_title('PageValues by Revenue')
sns.boxplot(x='Revenue', y='ExitRates', data=df, ax=axes[1])
axes[1].set_title('ExitRates by Revenue')
sns.boxplot(x='Revenue', y='ProductRelated_Duration', data=df, ax=axes[2])
axes[2].set_title('ProductRelated_Duration by Revenue')
plt.tight_layout()
plt.show()

# Correlation heatmap
numeric_df = df.select_dtypes(include=np.number)
plt.figure(figsize=(14, 10))
sns.heatmap(numeric_df.corr(), cmap='coolwarm', center=0)
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```
